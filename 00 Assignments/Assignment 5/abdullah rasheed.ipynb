# **Task 1 : Data Wrangling**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


# Loading Data Set 
# Data Set is only loaded once and used for both tasks

titanic = sns.load_dataset('titanic')
titanic

#Identify Missing Values
#titanic.shape
#titanic.isnull().sum()

```

> **Replace Numeric Values With Mean**

```python
# We only have one numerical value with null entries (Age)
repl_age = titanic['age'].median()
repl_age
titanic['age'].fillna(repl_age,inplace=True)
titanic.head(10)

#titanic.isnull().sum() to recheck 
```

> **For Categorical Columns Use Mode**

```python
#For Deck
mode_val = titanic['deck'].mode()[0] # to find first most occuring value because there can be multiple values of same occurences
mode_val
titanic['deck'].fillna(mode_val,inplace=True)
titanic.isnull().sum()

#For Embarked
mode_val = titanic['embarked'].mode()[0] # to find first most occuring value because there can be multiple values of same occurences
mode_val
titanic['embarked'].fillna(mode_val,inplace=True)
titanic.isnull().sum()

#For Embarked Town
mode_val = titanic['embark_town'].mode()[0] # to find first most occuring value because there can be multiple values of same occurences
mode_val
titanic['embark_town'].fillna(mode_val,inplace=True)
titanic.isnull().sum()
```

> **Identify & Remove Outliners For Age**

```python
# To Visually See Outliners
#sns.boxplot(titanic['age'])

#IQR Method
Q1= titanic['age'].quantile(0.25)
Q3 = titanic['age'].quantile(0.75)
IQR = Q3-Q1
upper_limit = Q3 + (1.5*IQR)
lower_limit = Q1 - (1.5*IQR)

titanic_upper = titanic[titanic['age'] > upper_limit]
#titanic_upper
titanic_lower = titanic[titanic['age']<lower_limit]
#titanic_lower
titanic['age'] = titanic['age'].replace(titanic_upper.values,repl_age)
titanic['age'] = titanic['age'].replace(titanic_lower.values,repl_age)
titanic.head(20)
```

> **Identify And Remove Outliners For Fare**

```python
Q1 = titanic['fare'].quantile(0.25)
Q3 = titanic['fare'].quantile(0.75)
IQR = Q3 - Q1

upper_limit = Q3 + (1.5 * IQR)
lower_limit = Q1 - (1.5 * IQR)

avg_fare = titanic['fare'].median()

titanic_upper = titanic.loc[titanic['fare'] > upper_limit, 'fare']
titanic_lower = titanic.loc[titanic['fare'] < lower_limit, 'fare']

titanic['fare'] = titanic['fare'].replace(titanic_upper.values, avg_fare)
titanic['fare'] = titanic['fare'].replace(titanic_lower.values, avg_fare)

titanic.head(30)
```

> **Feature Transformation**

```python
family_size=titanic[["sibsp","parch"]]
family_size
```

```python
age_range = [0.42, 18.00, 40.66, 80.00]
age_groups = ["Child", "Adult", "Senior"]

titanic['age_group'] = pd.cut(titanic['age'], bins=age_range, labels=age_groups)

age_feature=titanic[['age','age_group']]

#titanic.head(10)
age_feature.head(10)
```

>> **Catogorical Data Encoding**

```python
titanic['sex'] = titanic['sex'].replace({'male':0,'female':1})
titanic.head(10)
```

>> **Normalization**

```python
titanic['fare'] = (titanic['fare']-titanic['fare'].min())/(titanic['fare'].max()-titanic['fare'].min())
titanic['age'] = (titanic['age']-titanic['age'].min())/(titanic['age'].max()-titanic['age'].min())
titanic.head(10)
```
