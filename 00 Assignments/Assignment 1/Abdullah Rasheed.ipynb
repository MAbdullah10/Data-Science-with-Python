# Check Age and Name and Print Number

```python
name = input("Enter Your Name: ")
age = int(input("Enter Your Age: "))
num = int(input("Enter A Number Between 1 and 100: "))
def calculatesum(n):
    i = 1
    sum = 0
    while i <= 100:
        if n % i == 0:
            sum = sum+i 
        i = i+1
    return sum
    
if age < 0:
    print("Error")
    quit
elif age > 0 and age >= 18:
    print('Welcome ', name)
    print("You are eligible to vote")
    print("Sum of numbers divisble by ",num, "is",calculatesum(n=num))
    
else:
    print('Welcome ', name)
    print("You are not eligible to vote")
    print("Sum of numbers divisble by  ",num, "is ",calculatesum(n=num))
```
