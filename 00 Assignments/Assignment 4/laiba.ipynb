Task 1: Matplotlib - Basic Plots
Use the Titanic dataset to create:
i   A line plot showing the number of passengers over different ages.
ii  A bar chart comparing the number of survivors in each class.
iii A scatter plot between age and fare.
(Customize the plots with labels, titles, legends, and colors)

```python
import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset("titanic")
titanic
age_counts = titanic['age'].value_counts().sort_index()

plt.figure(figsize=(12, 6))
plt.plot(age_counts.index, age_counts.values, marker='o', color='lightpink', linewidth=1)

plt.title('Passengers by Age', fontsize=16)
plt.xlabel('Age', fontsize=14)
plt.ylabel('Passengers', fontsize=14)
plt.grid()
plt.xticks(range(0, 81, 5))  
plt.xlim(0, 80)  
plt.show()
```

```python
import seaborn as sb
import matplotlib.pyplot as mb


Titanic = sb.load_dataset("titanic")

mb.figure(figsize=(12, 6))
sb.barplot(x="pclass", y="survived", hue="sex", data=Titanic, palette="coolwarm")

mb.title('Survivors by Class and Gender', fontsize=16)
mb.xlabel('Class', fontsize=14)
mb.ylabel('Survivors', fontsize=14)
mb.legend(title='Gender')
mb.grid()
mb.show()

```

Task 2: Seaborn - Statistical Data Visualization
Using the Iris dataset, create:
i   A histogram for petal length distribution.
ii  A box plot comparing petal widths across species.
iii A heatmap for correlation between numerical features.
(Apply different color palettes and styles)

```python
import seaborn as sb
import matplotlib.pyplot as plt

Iris = sb.load_dataset("iris")

plt.figure(figsize=(10, 6))
sb.histplot(Iris['petal_length'], bins=20, kde=True, color='skyblue', edgecolor='darkblue')
plt.title('Petal Length Distribution', fontsize=16)
plt.xlabel('Petal Length', fontsize=14)
plt.ylabel('Frequency', fontsize=14)
plt.show()
```

```python
import seaborn as sb
import matplotlib.pyplot as plt

Iris = sb.load_dataset("iris")
plt.figure(figsize=(10, 6))
sb.boxplot(x='species', y='petal_width', data=Iris, palette='pastel')
plt.title('Petal Width Across Species', fontsize=16)
plt.xlabel('Species', fontsize=14)
plt.ylabel('Petal Width', fontsize=14)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

iris_data = sns.load_dataset('iris')

correlation_matrix = iris_data.corr(numeric_only=True)

plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation of Iris Dataset Features', fontsize=16)
plt.show()
```

Task 3: Plotly - Interactive Visualizations
Using the Titanic dataset:
i   Create an interactive line chart showing survival rates across age groups.
ii  Plot a 3D scatter plot for age, fare, and survival status.
iii Design an interactive pie chart showing the distribution of passenger classes.
(Ensure hover effects and legends are included)

```python
import pandas as pd
import plotly.express as px
import seaborn as sns

titanic = sns.load_dataset("titanic")

titanic['age_group'] = pd.cut(titanic['age'], bins=[0, 18, 30, 50, 80], labels=['0-18', '19-30', '31-50', '51-80'])

survival_rate = titanic.groupby('age_group')['survived'].mean().reset_index()

fig = px.line(
    survival_rate,
    x='age_group',
    y='survived',
    title='Survival Rates Across Age Groups',
    labels={'survived': 'Survival Rate', 'age_group': 'Age Group'},
    markers=True
)

fig.update_traces(line=dict(color='black'))
fig.show()
```

```python
import plotly.express as px
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

plt.figure(figsize=(12,8))
fig2 = px.scatter_3d(titanic, x='age', y='fare', z='survived',
                      color='survived',
                      title='3D Scatter Plot of Age, Fare and Survival Status',
                      labels={'survived': 'Survival Status'})
fig2.show()

```

```python
import plotly.express as px
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

pie = px.pie(titanic, names='class', title='Distribution of Passenger Classes',
               color_discrete_sequence=px.colors.sequential.Pinkyl_r)
pie.update_traces(textinfo='percent+label')  
pie.show()
```
