# **Task 1**
> **Matplotlib** 

> **Line Graph To Compare No of Passengers Of Different Ages**

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

titanic = sns.load_dataset("titanic")
titanic
age = np.array(titanic['age'])
plt.figure(figsize=(60,8),facecolor='LightBlue')
plt.plot(age,color='white',linewidth=3.5,marker="o",label='Ages') #We only give age in plot() because if we dont specify any creteria else it will Take No Of Rows By Default because each row represent a passenger
plt.legend(fontsize=20)
plt.title('Representation Of No Of Passengers With Respect To Different Ages',fontweight='bold',fontsize=28)
plt.xlabel('No Of Passengers',fontweight='bold',fontsize=20)
plt.xticks(fontsize=16,fontweight='bold')
plt.ylabel('Age Of Passenger',fontweight='bold',fontsize=20)
plt.yticks(fontsize=16,fontweight='bold')
#To Change BG Color Of Graph We Use .gca to get axis  (Research Done On Internet)
plt.gca().set_facecolor('black')
plt.grid()
plt.show()
```

> **Bar Graph To Compare No Of Survivors In Each Class**

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

titanic = sns.load_dataset("titanic")
titanic

survive= titanic[titanic['survived']==1]['class'].value_counts()
EachClass=survive.index

plt.figure(figsize=(10,8),facecolor='RoyalBlue')
plt.bar(EachClass,survive,color=['Purple','LightPink','Grey']) 
plt.title('Representation Of No Of Survivors In Each Class',fontweight='bold',fontsize=24)
plt.xlabel('Classes',fontweight='bold',fontsize=14)
plt.xticks(fontsize=16,fontweight='bold')
plt.ylabel('No Of Survivors',fontweight='bold',fontsize=14)
plt.yticks(fontsize=16,fontweight='bold')
plt.gca().set_facecolor('gold')
plt.show() 
```

> **Scatter Plot Between Age and Fare**

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
titanic = sns.load_dataset("titanic")
titanic
fare = np.array(titanic['fare'])
age = np.array(titanic['age'])
plt.figure(figsize=(20,8),facecolor='LightGrey')
plt.scatter(age,fare,color='black')
plt.title('Representation Of Fare With Respect To Different Ages',fontweight='bold',fontsize=24)
plt.xlabel('Age',fontweight='bold',fontsize=14)
plt.xticks(fontsize=16,fontweight='bold')
plt.ylabel('Fare',fontweight='bold',fontsize=14)
plt.yticks(fontsize=16,fontweight='bold')
plt.gca().set_facecolor('gold')
plt.grid()
plt.show() 
```

> **Histogram For Petal Distribution**

```python
import seaborn as sns
import matplotlib.pyplot as plt

petals = sns.load_dataset('iris')
petals
plt.figure(figsize=(20,8),facecolor='LightPink')
sns.set_palette('deep')
sns.histplot(data=petals,x='petal_length')

plt.title('Representation Of Distribution Of Petal Length',fontweight='bold',fontsize=24)
plt.xlabel('Petal Length',fontweight='bold',fontsize=14)
plt.xticks(fontsize=16,fontweight='bold')
plt.ylabel('Count',fontweight='bold',fontsize=14)
plt.xlim(1)
plt.yticks(fontsize=16,fontweight='bold')
plt.gca().set_facecolor('black')
plt.show()
```

> **Box Plot Of Petal Width Across Species**

```python
import seaborn as sns
import matplotlib.pyplot as plt

petals = sns.load_dataset('iris')
petals
sns.set_palette('muted')
plt.figure(figsize=(20,8),facecolor='DarkBlue')
sns.boxplot(x='species',y='petal_width',data=petals,hue='species',showmeans = True, meanprops= {
                                            "marker" : ">",
                                            "markersize" : "10",
                                            "markeredgecolor" : "white"
               },palette=['red','black','yellow'],saturation=0.5)

plt.title('Representation Of Petal Width According To Different Species',fontweight='bold',fontsize=24,color='white')
plt.xlabel('Species',fontweight='bold',fontsize=14,color='white')
plt.xticks(fontsize=16,fontweight='bold',color='white')
plt.ylabel('Petal Width',fontweight='bold',fontsize=14,color='white')

plt.yticks(fontsize=14,fontweight='bold',color='white')
plt.gca().set_facecolor('lightgrey')
plt.show()
```

> **Heat Map Between Coorelation Between Different Numerical Values**

```python
import seaborn as sns
import matplotlib.pyplot as plt

petals = sns.load_dataset('iris')
petals
numericFeatures = petals.select_dtypes(include=['number']) # To Include Only Numeric (Help Taken From AI)
correlation = numericFeatures.corr()
sns.set_palette('muted')
plt.figure(figsize=(20,8),facecolor='White')
sns.heatmap(correlation,annot=True,linewidths=2.5,cmap='magma')
plt.title('Representation Of Coorelation Of Numeric Values Only',fontweight='bold',fontsize=24,color='Black')
plt.show()
```

# **Task 3**
> **Plotly**

> **Line Graph Of Survival Rates Accros Age Groups**  

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
titanic = sns.load_dataset('titanic')
titanic
graph = px.line(titanic,x='age',y='survived',title='Line Graph To Represent Survival Rate Of Different Age Groups',labels={'age': 'Ages','survived':'Survival Rate'})
graph.update_layout(
    font_family="Times New",
    font_color="white",
    title_font_family="Times New Roman",
    title_font_color="White",
    paper_bgcolor='black',
    title_font_size=24,)
graph.show()
```

> **3-D Scatter Graph For Age, Fare and Survival Status**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
titanic = sns.load_dataset('titanic')
titanic
graph = px.scatter_3d(titanic,x='age',y='fare',z='survived',color='age',title='3-D Scatter Plot Between Fare, Age and Survival Status',labels={'fare':'Fares','survived':'Survival Status','age':'Age'})
graph.update_layout(
    font_family="Times New",
    font_color="white",
    title_font_family="Times New Roman",
    title_font_color="White",
    paper_bgcolor='black',
    title_font_size=24,
    legend_title_font_color="Blue")
graph.show()
```

> **Pie Chart Showing Distribution Of Passenger Classes**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import numpy as np
titanic = sns.load_dataset('titanic')
noOfPassengers = np.array(titanic["pclass"].value_counts())
pclasses = np.array(titanic["pclass"].value_counts().index)
pclasses
graph = px.pie(titanic,values=noOfPassengers,names=pclasses,labels={'names':'Pclass','values':'No Of Passengers'},title='Distribution Of No of Passengers Over Passenger Classes')
graph.update_layout(
    font_family="Times New",
    font_color="white",
    title_font_family="Times New Roman",
    title_font_color="White",
    paper_bgcolor='maroon',
    title_font_size=28,
    legend_title="PClass",
    legend_title_font_color="Gold")
graph.show()


```
