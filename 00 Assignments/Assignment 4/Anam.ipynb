#### Task 1: Matplotlib - Basic Plots
Use the Titanic dataset to create: <br>
i.   A line plot showing the number of passengers over different ages. <br>
ii.  A bar chart comparing the number of survivors in each class. <br>
iii. A scatter plot between age and fare. <br>
(Customize the plots with labels, titles, legends, and colors) 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

ship = sns.load_dataset("titanic")

plt.figure(figsize=(12,6)) 

age_count = ship["age"].value_counts()
age = ship["age"]

sns.lineplot(age_count, color="green", marker="o")
plt.plot()
plt.title("Agewise count of Passengers")
plt.xlabel("Age")
plt.ylabel("number of passengers")
plt.show()
```

```python
age_count
```

```python
plt.figure(figsize=(7.5,6)) 
survivors = ship["survived"].value_counts()
# pd.value_counts(ship['class']).plot.bar()
sns.barplot(y=survivors,x=ship["class"],color="purple", saturation=0.2)
plt.title("Number of survivors in each class")
plt.ylabel("No of survirors")
plt.plot()
plt.show()
```

```python
survivors
```

```python
plt.figure(figsize=(12,6)) 
plt.scatter(x="fare",y="age", data = ship, color="grey", marker= "x")
plt.title("Fare vs Age")
plt.ylabel("Age")
plt.xlabel("Fare")
plt.show()
```

#### Task 2: Seaborn - Statistical Data Visualization 
Using the Iris dataset, create: <br>
i   A histogram for petal length distribution. <br>
ii  A box plot comparing petal widths across species. <br>
iii A heatmap for correlation between numerical features. <br>
(Apply different color palettes and styles)

```python
import matplotlib.pyplot as plt
import seaborn as sns

flowers = sns.load_dataset("iris")
plt.figure(figsize=(10,5))
# sns.histplot(data = flowers, x = "petal_length", kde = True, hue = "species", palette="pastel")
sns.histplot(data=flowers, x = "petal_length" ,color='brown')
plt.title("Petal Length Distribution")
plt.ylabel("Frequency")
plt.xlabel("Petal Length")
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x="petal_width", y="species", data= flowers, color ='lightblue' ,
               showmeans = True, meanprops= {
                                            "marker" : "o",
                                            "markersize" : "5",
                                            "markeredgecolor" : "red",
                                            "markerfacecolor" : "black"
               })
plt.title("Petal width distribution among species", fontsize="16")
plt.ylabel("Species")
plt.xlabel("Petal Width")
```

```python
flowers
```

```python
numeric_data = flowers.select_dtypes(include=['float64', 'int64'])
corr_1 = numeric_data.corr()
corr_1
sns.heatmap(corr_1,annot=True)
```


#### Task 3: Plotly - Interactive Visualizations
Using the Titanic dataset: <br>
i   Create an interactive line chart showing survival rates across age groups. <br>
ii  Plot a 3D scatter plot for age, fare, and survival status. <br>
iii Design an interactive pie chart showing the distribution of passenger classes. <br>
(Ensure hover effects and legends are included) 

```python
import plotly.express as px

ship = sns.load_dataset("titanic")

bin_edges = [0, 12, 20, 50, 85]
age_groups = ["Child", "Teen", "Adult", "Senior"]

ship["age_group"] = pd.cut(ship['age'], bins=bin_edges, labels=age_groups)
survival_rate = ship.groupby("age_group")["survived"].mean()
survival_rate
fig = px.line(survival_rate, labels={'age_group': 'Age Group', 'value': 'Survival Rate'}, title='Survival Rate by Age Group')
fig.show()
```

```python
fig = px.scatter_3d(ship, x = 'age', y = 'fare', z = 'survived',color = 'class') 
 
fig.show()
```

```python
fig = px.pie(ship, names='pclass', title='Distribution of Passenger Classes', labels={'pclass': 'Passenger Class'})
fig.show()
```
