## Task 1
**Tuple Practice**

```python
#Creation of Tuple
tuple1 = ("Strawberry","Mango","Apple","Amrood","Orange")

# Printing of second and fourth item
print(tuple1[1])
print(tuple1[3])

#tuple1[3] = ("Chiku",) Error : Type tuple does not allow in line assignment

# Tuple to list
tuplelist = list(tuple1) 
print(tuplelist)

# Add new item to list
tuplelist.append("Peach")
print(tuplelist)

# List to tuple
tuple1 = tuple(tuplelist)
tuple1
```

## Task 2
**List Practice**

```python
# Creation of Random List
mylist = [2,3,9,4,7,1,7,2]
print(mylist)

#List Sort
mylist.sort()
print(mylist)

#Add Two Numbers
mylist.append(10)
mylist.append(1)
print(mylist)

#Remove First Occurance
mylist.remove(7)
print(mylist)

#Reverse List
mylist.reverse()
mylist
```

## Task 3
**Dictionary Practice**

```python
#Dictionary Creation
dictionary1 = {"Abdullah": 85, "Ali": 23, "Waleed": 55,"Faizan": 78,"Umair": 81}
print(dictionary1)

# Add new record
dictionary1.update({"Umer": 40})
print(dictionary1)

#Update Score of a student
dictionary1["Umair"] = 66
print(dictionary1)

#Remove One Student
dictionary1.pop("Umair")
print(dictionary1)

#Print Keys And Values Seperately
print(dictionary1.keys())
print(dictionary1.values())
```

## Task 4
**Set Practice**


```python
# Set of 6 unique colors
myset = {"Red","Yellow","Orange","Green","Pink"}
print(myset)

#Add two new colors
myset.add("Purple")
myset.add("Blue")
print(myset)

#Adding Duplicate Color
myset.add("Blue")
print(myset) #It Does not add it

#Remove Specific Color
myset.remove("Red")
print(myset)

#Union And Intersection
#Union Built in Method, Can also use | or update method
myset2 = {"Red","Grey","Black","Blue"}
myset3 = myset2.union(myset)
print("Union: ",myset3)

# Intersection using built in method
myset3 = myset2.intersection(myset)
print("Intersection: ",myset3)
```
