#### _Task 1: 1D Array - Sum of Even and Odd Indexed Elements_
You are given a 1D array of integers. Write a function that computes the sum of all elements at even indices and the sum of all elements at odd indices. <br>
Input: A 1D array of integers.
Example Input:[2, 5, 3, 8, 10, 7]
Expected Output: Sum of elements at even indices = 15  
		 Sum of elements at odd indices = 20
Hint: Remember, indices start from 0. Even indices are 0, 2, 4... Odd indices are 1, 3, 5...

```python
import numpy as np

num = np.array([4,3,8,7,2,10,1,12])

def array_sum(num):
    sum_even = 0
    sum_odd = 0
    i=0
    while(i<len(num)):
        if(i%2==0):
            sum_even = sum_even + num[i]
        else: 
            sum_odd = sum_odd + num[i]
        i+=1
    print("Sum of elements at even indices = ", sum_even)
    print("Sum of elements at odd indices = ", sum_odd)

array_sum(num)
```

#### _Task 2: 2D Array - Row-wise Maximum_
You are given a 2D array with each row containing scores of students in different subjects. Write a function that finds the highest score in each row. <br>
Input: A 2D array of student scores. <br>
Example Input:[ <br>
 [65, 78, 85, 92, 55],   # Student 1  <br>
 [82, 91, 88, 60, 70],   # Student 2  <br>
 [95, 80, 75, 88, 72]    # Student 3  <br>
]  <br>
Expected Output:[92, 91, 95] <br>
Hint: Use the built-in max() function to get the maximum score in each row.

```python
import numpy as np
scores = np.array([[65, 78, 85, 92, 55],   # Student 1 
 [82, 91, 88, 60, 70],   # Student 2 
 [95, 80, 75, 88, 72]    # Student 3
])

def highest_score(scores):
    shape = scores.shape 
    highest = np.empty(shape[0],dtype= int)
    #print(shape)
    i=0
    n=5
    while (i < shape[0]):  
        max_value = np.max(scores[i,:n])
        highest[i] = max_value       
        i += 1 
        n+=shape[1]
    print("Highest scores of each row ",highest)
highest_score(scores)
```
