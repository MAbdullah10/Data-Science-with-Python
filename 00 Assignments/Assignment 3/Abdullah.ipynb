# Task 1
> 1-D Array Sum of Evem And Odd Index Elements 

```python
import numpy as np

def sumOfIndex(array):
    sumEven=0
    sumOdd=0
    i=0
    while i < len(array):
        if i % 2 == 0:
            #It is even
            sumEven=sumEven+array[i]
        else:
            #It is Odd
            sumOdd=sumOdd+array[i]
        i = i+1
    
    print("Sum Of Even Indexed Numbers Is: ",sumEven)
    print("Sum Of Odd Indexed Numbers Is: ",sumOdd)

array = np.array([2,5,3,8,10,7])
print('Input Array: ',array)
sumOfIndex(array=array)
```

# Task 2
> Find Maximum Score In Each Row Of 2-D Array

```python
import numpy as np

def maxIneachRow(array):
    print('Highest Score In Each Row Is: ',array.max(axis=1)) # AXIS 1 TO GET ROW WISE . BY DEFAULT IT IS 0 MEANS COLUMN WISE

array = np.array([[65,78,85,92,55],
                  [82,91,88,60,70],
                  [95,80,75,88,72]])


print('Input Array: ',array)
maxIneachRow(array=array)
```
